Mit diesem Notebook können Sie die aus den Praktika bekannten Dateien "paketdaten.txt", "fliesskommawerte.txt", "bewegungen.txt" und "kant-vorrede.txt" nutzen. Sie finden unten ein Codebeispiel, das die Datei "kant-vorrede.txt" öffnet, einliest und zeilenweise ausgibt:


```python
# Einbinden des "OS"-Moduls (OS: Operating System (Betriebssystem))
import os

# listdir() gibt den Inhalt des aktuellen Verzeichnisses aus
# (das Verzeichnis, in dem sich die Jupyter-Notebook Datei befindet)
for eintrag in os.listdir():
        print(eintrag)


# Zwei Leerzeilen ausgeben
print("\n")

# Öffnen der Datei
datei = open("kant-vorrede.txt", "r")

# Zeilenweises ausgeben des Dateiinhalts.
for zeile in datei:
    print(zeile.strip())
```
